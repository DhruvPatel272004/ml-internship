```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
```

```python
dataset = pd.read_csv("../../Data/1) Iris.csv")
print(dataset.head())
```

```python
dataset.info()
```

```python
print(dataset.isna().sum())
```

```python
print(dataset.duplicated().sum())
```

```python
dataset = dataset.drop_duplicates()
```

```python
print(dataset.duplicated().sum())
```

```python
dataset.info()
```

dtypes: float64(4), object(1)

species column has datatype as object and considered as categorical value. 
need to convert the column to numerical type

```python
species_names = []
species_names = dataset['species'].unique()
print(species_names)
```

species have three unique values, so we can map them as 1 - setosa,2 - versicolor and 3 - virginica

```python
dataset['species'] = dataset['species'].map({'setosa':1 , 'versicolor':2, 'virginica':3})
```

```python
print(dataset.head())
```

```python
dataset.info()
```

```python
dataset.shape
```

```python
data = dataset.values
X = data[:,0:4]
y = data[:,4]

print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1, stratify=y)
```

```python
print(X_train.shape,X_test.shape)
print(y_train.shape,y_test.shape)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
print(species_names)
```

```python
for k in range(1,11):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train,y_train)
    y_pred = knn.predict(X_test)
    print("K=", k)
    print(f"Accuracy:{(knn.score(X_test,y_test) * 100):.2f}%")
    report = classification_report(y_test, y_pred, target_names=species_names)
    print(report)
    cm = confusion_matrix(y_test, y_pred, labels=dataset['species'].unique())
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    disp.plot()
    plt.show()
    print("="*200)
```

The accuracy value for k=1 93.33% and rest of ks from 2 to 11 have same accuracy of 96.67%.
