# Categorical Cols

Categorical cols - `State`, `International plan`,`Voice mail plan` and `Churn`.

Columns like `International plan` and `Voice mail plan` has `Yes` and `No` as values which is object type which gets converted to `1` for `Yes` and `0` for `No`.

```python
# Converting categorical data to numerical data
dataset['International plan'] = dataset['International plan'].map({'Yes': 1, 'No': 0})
dataset['Voice mail plan'] = dataset['Voice mail plan'].map({'Yes': 1, 'No': 0})

print(dataset[['International plan','Voice mail plan']].head())
```

`Churn` column is boolean which can be replaced by 0 and 1.

```python
# Converting target column from boolean to int data type
dataset['Churn'] = dataset['Churn'].astype(int)
print(dataset[['Churn']].head())
```

`State` column is unable to represent in order and has multiple entries which leads to `one-hot encoding`.

```python
# One hot encoding for 'State' column and converting that to into int data type
dataset = pd.get_dummies(dataset, columns=['State'], drop_first=True).astype(int)
```

# Features Selection

Rearranging columns to seperate features easily

```python
# Rearranging columns
cols = ["Churn"]
dataset = dataset[cols + [col for col in dataset.columns if col not in cols]]
print(dataset.head())
```

```python
# Conerting data ready for features selection
data = dataset.values
```

```python
# Features selection
X = data[:,1:]
y = data[:, 0]
```

Spliting of Dataset - Train/Test set

```python
# Spliting of data into training and testing test
# 80/20 ratio
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

Feature scaling

```python
# Feature scaling
scaler = StandardScaler()

# Fitting and scaling X_train
X_train = scaler.fit_transform(X_train)

# Scaling X_test
X_test = scaler.transform(X_test)
```

```python
# Printing shapes for splited data to confirm the spliting

print("X_train shape = ", X_train.shape)
print("X_test shape = ", X_test.shape)
print("y_train shape = ", y_train.shape)
print("y_test shape = ", y_test.shape)
```
